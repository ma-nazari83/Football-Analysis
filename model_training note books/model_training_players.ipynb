```python
from pathlib import Path
import numpy as np
from ultralytics import YOLO
import torch
DEVICE = "cpu"
ROOT = Path(".").resolve()
DATASET_DIR = ROOT / "datasets/players"

print("Project root:", ROOT)
print("Device:", DEVICE)
```

```python
print(np.__version__)
print(hasattr(np, "trapz"))
```

```python
DATASET_DIR
```

```python
yaml_path = DATASET_DIR/"data.yaml"
```

```python
original_load = torch.load
def patched_load(*args, **kwargs):
    kwargs['weights_only'] = False
    return original_load(*args, **kwargs)
torch.load = patched_load
```

```python
model  = YOLO("yolov8n.pt")
```

```python
metrics = model.train(
    data=str(yaml_path),
    epochs=50,
    imgsz=640,
    batch=4,
    device=DEVICE,
    workers=2,  # set to 2–4 on Linux/macOS if stable
    project=str(ROOT / "runs"),
    name="detect_custom",
    exist_ok=True,
)
```

```python
run_dir = ROOT / "runs" / "detect_custom"
best_weights = run_dir / "weights" / "best.pt"

val_model = YOLO(str(best_weights))
metrics = val_model.val(data=str(yaml_path), device=DEVICE, workers=2)
print(metrics)
```

```python
PREDICT_SOURCE = str(DATASET_DIR / "images" / "val")  # folder of .jpg images

pred_model = YOLO(str(best_weights))
pred_model.predict(
    source=PREDICT_SOURCE,
    device=DEVICE,
    conf=0.25,
    save=True,
    project=str(ROOT / "runs"),
    name="predict_custom",
    exist_ok=True,
)
```
